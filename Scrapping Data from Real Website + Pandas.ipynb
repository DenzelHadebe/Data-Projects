```python
from bs4 import BeautifulSoup
import requests
```

```python
url = 'https://en.wikipedia.org/wiki/List_of_largest_companies_in_South_Africa'

# Add headers to mimic a browser request
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
    'Accept': 'text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8',
    'Accept-Language': 'en-US,en;q=0.5',
    'Connection': 'keep-alive',
    'Upgrade-Insecure-Requests': '1',
}

page = requests.get(url, headers=headers)

# If successful, parse the page
if page.status_code == 200:
    soup = BeautifulSoup(page.content, 'html.parser')
else:
    print(f"Failed to retrieve the page: {page.status_code}")
```

```python
print(soup)
```

```python
table = soup.find_all('table')[0]
```

```python
world_titles = table.find_all('th')
```

```python
world_titles
```

```python
world_table_titles = [title.text.strip() for title in world_titles]
print(world_table_titles)
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns = world_table_titles)
df
```

```python
column_data = table.find_all('tr')
```

```python
for row in column_data[1:] :
    row_data = row.find_all('td')
    individual_data = [data.text.strip() for data in row_data]
    length = len(df)
    df.loc[length]= individual_data
   # print(individual_data) 
```
